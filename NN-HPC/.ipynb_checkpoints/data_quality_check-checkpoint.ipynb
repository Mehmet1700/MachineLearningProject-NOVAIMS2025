```python
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.pipeline import Pipeline

# Ensure we can import from local modules
if '.' not in sys.path:
    sys.path.append('.')

from data_loaders.cars_data import load_full_train_and_test
# New imports for the modular pipeline
from utils.imputation import get_imputation_pipeline, get_encoding_transformer, get_scaler, drop_interval_cols
from utils.feature_engineering import FeatureEngineeringTransformer, load_mappings

%matplotlib inline
```

## 1. Load Data (Cleaning & Mapping)

```python
train_path = "../data/train.csv"
test_path = "../data/test.csv"
mapping_dir = "../mapping"

print("Loading data...")
X_train, y_train, X_test, test_ids = load_full_train_and_test(
    train_path=train_path,
    test_path=test_path,
    mapping_dir=mapping_dir,
    return_test_ids=True,
    debug_cleaning=True  # Enable detailed cleaning logs
)

print(f"Train shape: {X_train.shape}")
print(f"Test shape:  {X_test.shape}")
```

## 2. Inspect Missing Values (Pre-Imputation)

```python
def plot_missing(df, title):
    missing = df.isnull().sum()
    missing = missing[missing > 0].sort_values(ascending=False)
    if missing.empty:
        print(f"No missing values in {title}")
        return
    
    plt.figure(figsize=(10, 6))
    sns.barplot(x=missing.values, y=missing.index)
    plt.title(f"Missing Values in {title}")
    plt.xlabel("Count")
    plt.show()
    return missing

print("--- Train Missing ---")
miss_train = plot_missing(X_train, "Train Data")
print(miss_train)

print("\n--- Test Missing ---")
miss_test = plot_missing(X_test, "Test Data")
print(miss_test)
```

```python
# Check Target Distribution
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.histplot(y_train, kde=True)
plt.title("Price Distribution")

plt.subplot(1, 2, 2)
sns.histplot(np.log1p(y_train), kde=True)
plt.title("Log(Price) Distribution")
plt.show()
```

## 4. Apply Preprocessing (New Modular Pipeline)
We verify that the new modular pipeline (Imputation -> Feature Engineering -> Encoding -> Scaling) works correctly and produces valid output.

```python
# Helper functions for debugging
def check_validity(X, name):
    if hasattr(X, "toarray"):
        X = X.toarray()
        
    # Handle Pandas DataFrame
    if hasattr(X, "columns"):
        nans = X.isna().sum().sum()
        # Check infs only on numeric columns
        numeric_cols = X.select_dtypes(include=[np.number])
        if not numeric_cols.empty:
            infs = np.isinf(numeric_cols).sum().sum()
        else:
            infs = 0
    else:
        # Handle Numpy Array
        X = np.array(X)
        if np.issubdtype(X.dtype, np.number):
            nans = np.isnan(X).sum()
            infs = np.isinf(X).sum()
        else:
            # Object array (strings etc)
            nans = pd.isna(X).sum()
            infs = 0

    print(f"{name}: NaNs={nans}, Infs={infs}")
    if nans > 0:
        print(f"WARNING: NaNs found in {name}!")

def show_nan_cols(df, name):
    if not hasattr(df, "columns"):
        return
    nans = df.isna().sum()
    nans = nans[nans > 0]
    if not nans.empty:
        print(f"\n[DEBUG] {name} NaN columns:\n{nans}")
    else:
        print(f"\n[DEBUG] {name} has NO NaNs.")

# 1. Load Mappings
brand_map, model_map = load_mappings(mapping_dir)

# --- STEP 1: IMPUTATION ---
print("\n=== STEP 1: IMPUTATION ===")
imputation_pipe = get_imputation_pipeline()

# Fit on Train
print("Fitting Imputer on Train...")
imputation_pipe.fit(X_train, y_train)

# Transform
X_train_imp = imputation_pipe.transform(X_train)
X_test_imp = imputation_pipe.transform(X_test)

# Debug
print(f"Imputed Train Shape: {X_train_imp.shape}")
check_validity(X_train_imp, "Train Imputed")
check_validity(X_test_imp, "Test Imputed")

show_nan_cols(X_train_imp, "Train Imputed")
show_nan_cols(X_test_imp, "Test Imputed")
```

```python
# --- STEP 2: FEATURE ENGINEERING ---
print("\n=== STEP 2: FEATURE ENGINEERING ===")
fe_transformer = FeatureEngineeringTransformer(brand_mapping=brand_map, model_mapping=model_map)

# Fit on Train (Imputed)
print("Fitting FE on Train Imputed...")
fe_transformer.fit(X_train_imp, y_train)

# Transform
X_train_fe = fe_transformer.transform(X_train_imp)
X_test_fe = fe_transformer.transform(X_test_imp)

# Debug
print(f"FE Train Shape: {X_train_fe.shape}")
check_validity(X_train_fe, "Train FE")
check_validity(X_test_fe, "Test FE")

show_nan_cols(X_train_fe, "Train FE")
show_nan_cols(X_test_fe, "Test FE")
```

```python
# --- STEP 3: ENCODING ---
print("\n=== STEP 3: ENCODING ===")
encoding_transformer = get_encoding_transformer()

# Fit on Train (FE)
print("Fitting Encoder on Train FE...")
encoding_transformer.fit(X_train_fe)

# Transform
X_train_enc = encoding_transformer.transform(X_train_fe)
X_test_enc = encoding_transformer.transform(X_test_fe)

# Get feature names for better debugging
try:
    feat_names = encoding_transformer.get_feature_names_out()
    X_train_enc_df = pd.DataFrame(X_train_enc, columns=feat_names, index=X_train_fe.index)
    X_test_enc_df = pd.DataFrame(X_test_enc, columns=feat_names, index=X_test_fe.index)
    print("Converted encoded arrays back to DataFrames for debugging.")
except Exception as e:
    print(f"Could not convert to DF: {e}")
    X_train_enc_df = X_train_enc
    X_test_enc_df = X_test_enc

# Debug
print(f"Encoded Train Shape: {X_train_enc_df.shape}")
check_validity(X_train_enc_df, "Train Encoded")
check_validity(X_test_enc_df, "Test Encoded")

if hasattr(X_train_enc_df, "isna"):
    show_nan_cols(X_train_enc_df, "Train Encoded")
    show_nan_cols(X_test_enc_df, "Test Encoded")
```

```python
# --- STEP 4: SCALING ---
print("\n=== STEP 4: SCALING ===")
# First drop any interval columns that might have leaked
X_train_enc_df = drop_interval_cols(X_train_enc_df)
X_test_enc_df = drop_interval_cols(X_test_enc_df)

scaler = get_scaler()

# Fit on Train (Encoded)
print("Fitting Scaler on Train Encoded...")
scaler.fit(X_train_enc_df)

# Transform
X_train_scaled = scaler.transform(X_train_enc_df)
X_test_scaled = scaler.transform(X_test_enc_df)

# Final Result
X_train_processed = X_train_scaled
X_test_processed = X_test_scaled

# Debug
print(f"Scaled Train Shape: {X_train_processed.shape}")
check_validity(X_train_processed, "Train Scaled")
check_validity(X_test_processed, "Test Scaled")
```

```python
# Check for NaNs in processed data
def check_validity(X, name):
    if hasattr(X, "toarray"):
        X = X.toarray()
        
    # Handle Pandas DataFrame
    if hasattr(X, "columns"):
        nans = X.isna().sum().sum()
        # Check infs only on numeric columns
        numeric_cols = X.select_dtypes(include=[np.number])
        if not numeric_cols.empty:
            infs = np.isinf(numeric_cols).sum().sum()
        else:
            infs = 0
    else:
        # Handle Numpy Array
        X = np.array(X)
        if np.issubdtype(X.dtype, np.number):
            nans = np.isnan(X).sum()
            infs = np.isinf(X).sum()
        else:
            # Object array (strings etc)
            nans = pd.isna(X).sum()
            infs = 0

    print(f"{name}: NaNs={nans}, Infs={infs}")
    if nans > 0:
        print(f"WARNING: NaNs found in {name}!")

check_validity(X_train_processed, "Train Processed")
check_validity(X_test_processed, "Test Processed")
```

```python
# Check the NaN Values in the processed data (which columns have NaNs)
print("\n--- Checking Processed Train Data ---")
miss_train_processed = pd.DataFrame(X_train_processed).isnull().sum()
miss_train_processed = miss_train_processed[miss_train_processed > 0]
if miss_train_processed.empty:
    print("No missing values in processed Train data.")
else:
    print(miss_train_processed)
```

```python
# Feature Names
try:
    # In the new modular pipeline, the encoding step defines the feature names
    feature_names = encoding_transformer.get_feature_names_out()
    print(f"Total Features: {len(feature_names)}")
    print("First 20 features:", feature_names[:20])
except NameError:
    print("Error: 'encoding_transformer' is not defined. Please run the 'STEP 3: ENCODING' cell first.")
except Exception as e:
    print("Could not retrieve feature names:", e)
```

```python
# --- STEP 5: FEATURE SELECTION (Filter Methods) ---
print("\n=== STEP 5: FEATURE SELECTION (Filter Methods) ===")
from utils.feature_selection import drop_low_variance_features, drop_highly_correlated_features

# We work on the processed (scaled) data
# Convert to DataFrame if it's a numpy array (which it is after scaling)
if not isinstance(X_train_processed, pd.DataFrame):
    X_train_fs = pd.DataFrame(X_train_processed, columns=feature_names, index=X_train.index)
    X_test_fs = pd.DataFrame(X_test_processed, columns=feature_names, index=X_test.index)
else:
    X_train_fs = X_train_processed.copy()
    X_test_fs = X_test_processed.copy()

print(f"Initial Feature Count: {X_train_fs.shape[1]}")

# 1. Variance Threshold
X_train_fs = drop_low_variance_features(X_train_fs, threshold=0.01)
# Apply same drop to test
remaining_cols = X_train_fs.columns
X_test_fs = X_test_fs[remaining_cols]
print(f"Features after Variance Filter: {X_train_fs.shape[1]}")

# 2. Correlation Filter
X_train_fs = drop_highly_correlated_features(X_train_fs, threshold=0.9)
# Apply same drop to test
remaining_cols = X_train_fs.columns
X_test_fs = X_test_fs[remaining_cols]
print(f"Features after Correlation Filter: {X_train_fs.shape[1]}")

# Debug
check_validity(X_train_fs, "Train FS")
check_validity(X_test_fs, "Test FS")
```

```python
# --- STEP 6: FEATURE SELECTION (Wrapper Methods - RFE) ---
print("\n=== STEP 6: FEATURE SELECTION (Wrapper Methods - RFE) ===")
import importlib
import utils.feature_selection
importlib.reload(utils.feature_selection)
from utils.feature_selection import perform_rfe_selection

# We continue with the features selected by the filter methods (X_train_fs)
# to save computation time.

# For this test, we select top 20 features
# We use a larger step to make it faster for the check
X_train_rfe = perform_rfe_selection(X_train_fs, y_train, n_features_to_select=20, step=5)

# Apply selection to test
selected_features_rfe = X_train_rfe.columns
X_test_rfe = X_test_fs[selected_features_rfe]

print(f"Features after RFE: {X_train_rfe.shape[1]}")
print("Selected Features:", list(selected_features_rfe))

# Debug
check_validity(X_train_rfe, "Train RFE")
check_validity(X_test_rfe, "Test RFE")
```
