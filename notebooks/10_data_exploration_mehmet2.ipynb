- Canonical Mapping and Normalization for model names

```python
import re
import pandas as pd
import numpy as np
import json

def norm_model(x: object) -> str | float:
    """Lowercase, keep [a-z0-9+], strip spaces/hyphens/underscores."""
    if pd.isna(x):
        return np.nan
    s = str(x).lower()
    s = re.sub(r'[\s\-_]+', '', s)           # drop spaces, hyphens, underscores
    s = re.sub(r'[^a-z0-9\+]', '', s)        # keep letters, digits, plus
    return s or np.nan
```

```python
# Load mapping JSON
with open("../mapping/modelname_mapping.json", "r", encoding="utf-8") as f:
    MM = json.load(f)

ALIASES = MM.get("aliases", {})
REGEX_RULES = MM.get("regex_rules", [])
```

```python
def apply_regex_rules(norm_key: str) -> str:
    for rule in REGEX_RULES:
        pat = rule["pattern"]
        rep = rule["replace"]
        norm_key = re.sub(pat, rep, norm_key)
    return norm_key
```

```python
def clean_model(series: pd.Series) -> pd.Series:
    normed = series.apply(norm_model)
    normed = normed.apply(lambda x: apply_regex_rules(x) if isinstance(x, str) else x)
    return normed.map(ALIASES)
```

```python
# Apply the clean model function to the 'model' column for train and test
x_train['model'] = clean_model(x_train['model'])
x_test['model'] = clean_model(x_test['model'])

# Apply the apply_regex_rules function to the 'model' column
x_train['model'] = x_train['model'].apply(lambda x: apply_regex_rules(x) if isinstance(x, str) else x)
x_test['model'] = x_test['model'].apply(lambda x: apply_regex_rules(x) if isinstance(x, str) else x)

# unique models in the column model for x_train
unique_models = x_train['model'].unique()
print("\nUnique models in 'model':\n", unique_models)
# print the number of unique models
print("\nNumber of unique models in 'model':", len(unique_models))
```

```python
import pandas as pd
import numpy as np
import re

# Define a normalization function
def clean_transmission(x):
    if pd.isna(x):
        return np.nan
    s = str(x).strip().lower()

    # handle common patterns
    if re.search(r"man", s):          # e.g. 'manual', 'manua', 'manuaL '
        return "manual"
    elif re.search(r"semi", s):       # e.g. 'semi-auto', 'semi-aut'
        return "semi-auto"
    elif re.search(r"auto", s):       # e.g. 'automatic', 'automati', 'utomatic'
        return "automatic"
    elif re.search(r"unk", s):        # e.g. 'unknown', 'unknow', 'nknown'
        return "unknown"
    elif re.search(r"oth", s):        # e.g. 'other'
        return "other"
    else:
        return np.nan   # if none matches, mark as missing

# Apply it to train and test
x_train["transmission"] = x_train["transmission"].apply(clean_transmission)
x_test["transmission"] = x_test["transmission"].apply(clean_transmission)

# Optional: check results
print(x_train["transmission"].value_counts(dropna=False))
print(x_test["transmission"].value_counts(dropna=False))
```

```python

def clean_fuel_type(x):
    if pd.isna(x):
        return np.nan
    s = str(x).strip().lower()

    # Match the main types
    if re.search(r"pet", s):          # petrol, petro, etrol, etc.
        return "petrol"
    elif re.search(r"die", s):        # diesel, diese, iese, etc.
        return "diesel"
    elif re.search(r"hyb", s):        # hybrid, hybr, ybrid, etc.
        return "hybrid"
    elif re.search(r"elec", s):       # electric
        return "electric"
    elif re.search(r"oth", s):        # other, othe, ther
        return "other"
    elif re.search(r"unk", s):        # if any 'unknown' appears
        return "unknown"
    else:
        return np.nan   # not recognized

# Apply cleaning to x_train and x_test
x_train["fuelType"] = x_train["fuelType"].apply(clean_fuel_type)
# Check results
print(x_train["fuelType"].value_counts(dropna=False))

x_test["fuelType"] = x_test["fuelType"].apply(clean_fuel_type)
# Check results
print(x_test["fuelType"].value_counts(dropna=False))
```

- So we have a few entries with year > 2020 and we have some missing values in year.
- We will replace the wrong and missing values with the median of the respective brand/model.
- We will also convert the year column to integer type.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer

# Columns to use
cols = ['Brand', 'model', 'mileage', 'year', 'previousOwners']
df_subset = x_train[cols].copy()

# We'll only impute where year is NaN or > 2020
mask_to_impute = df_subset['year'].isna() | (df_subset['year'] > 2020)

# Keep original year for safety
df_subset['year_original'] = df_subset['year']

# Define columns for preprocessing
cat_cols = ['Brand', 'model']
num_cols = ['mileage', 'year', 'previousOwners']  # 'year' included so KNN can impute it

preprocessor = ColumnTransformer([
    ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols),
    ('num_scaled', StandardScaler(), ['mileage', 'previousOwners']),
    ('year_passthrough', 'passthrough', ['year'])
])

# KNN Imputation pipeline
knn_pipeline = Pipeline([
    ('preprocess', preprocessor),
    ('impute', KNNImputer(n_neighbors=5, weights='distance'))
])

# Fit on the whole subset (KNN needs neighbors), then transform
imputed_array = knn_pipeline.fit_transform(df_subset[cat_cols + num_cols])

# The imputed 'year' is the LAST column in the transformed numeric block
imputed_years = imputed_array[:, -1]

# Build result and only overwrite where desired
df_imputed = df_subset.copy()
df_imputed.loc[mask_to_impute, 'year'] = imputed_years[mask_to_impute.values]

# Postprocess/guards
df_imputed['year'] = (
    df_imputed['year']
      .clip(lower=1980, upper=2020)
      .round()
      .astype(int)
)

# Optional: add a flag for transparency/debugging
df_imputed['year_imputed_knn'] = mask_to_impute.astype(int)

print(df_imputed[['Brand','model','mileage','previousOwners','year_original','year_imputed_knn','year']].head())
```
