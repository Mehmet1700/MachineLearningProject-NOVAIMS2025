```python
import re
import pandas as pd
import numpy as np
import json

def norm_model(x: object) -> str | float:
    """Lowercase, keep [a-z0-9+], strip spaces/hyphens/underscores."""
    if pd.isna(x):
        return np.nan
    s = str(x).lower()
    s = re.sub(r'[\s\-_]+', '', s)           # drop spaces, hyphens, underscores
    s = re.sub(r'[^a-z0-9\+]', '', s)        # keep letters, digits, plus
    return s or np.nan
```

```python
# Load mapping JSON
with open("../mapping/modelname_mapping.json", "r", encoding="utf-8") as f:
    MM = json.load(f)

ALIASES = MM.get("aliases", {})
REGEX_RULES = MM.get("regex_rules", [])
```

```python
def apply_regex_rules(norm_key: str) -> str:
    for rule in REGEX_RULES:
        pat = rule["pattern"]
        rep = rule["replace"]
        norm_key = re.sub(pat, rep, norm_key)
    return norm_key
```

```python
def clean_model(series: pd.Series) -> pd.Series:
    normed = series.apply(norm_model)
    normed = normed.apply(lambda x: apply_regex_rules(x) if isinstance(x, str) else x)
    return normed.map(ALIASES)
```

```python
# Apply the clean model function to the 'model' column
X['model'] = clean_model(X['model'])

# Apply the apply_regex_rules function to the 'model' column
X['model'] = X['model'].apply(lambda x: apply_regex_rules(x) if isinstance(x, str) else x)

# unique models in the column model
unique_models = X['model'].unique()
print("\nUnique models in 'model':\n", unique_models)
# print the number of unique models
print("\nNumber of unique models in 'model':", len(unique_models))
```

```python
# -----------------------------
# 1. Semantically encode fuelType
# -----------------------------
# Define your own mapping so similar types are closer numerically
fuel_mapping = {
    'Petrol': 0,
    'Diesel': 1,
    'Hybrid': 2,
    'Other': 3
}

# Apply the mapping
X['fuelType_encoded'] = X['fuelType'].map(fuel_mapping)

# -----------------------------
# 2. Preprocess features
# -----------------------------
categorical_cols = ['Brand', 'model', 'transmission', 'hasDamage']
numerical_cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize', 'paintQuality%', 'previousOwners']

preprocessor = ColumnTransformer([
    ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols),
    ('num', StandardScaler(), numerical_cols)
])

X_features = preprocessor.fit_transform(X)

# -----------------------------
# 3. Combine features with fuelType_encoded (keep NaNs!)
# -----------------------------
X_for_imputation = np.hstack([X_features.toarray(), X[['fuelType_encoded']].values])

# -----------------------------
# 4. Apply KNN imputer
# -----------------------------
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=5)
X_imputed = imputer.fit_transform(X_for_imputation)

# -----------------------------
# 5. Recover fuelType (only replace missing ones)
# -----------------------------
fuel_imputed = np.round(X_imputed[:, -1]).astype(int)

# Reverse the mapping
inv_fuel_mapping = {v: k for k, v in fuel_mapping.items()}

# Create a copy to store the filled version
X['fuelType_imputed'] = X['fuelType']

# Identify which rows were originally missing
missing_mask = X['fuelType_encoded'].isna()

# Replace only missing ones
X.loc[missing_mask, 'fuelType_imputed'] = pd.Series(
    fuel_imputed[missing_mask]
).map(inv_fuel_mapping).values

# -----------------------------
# 6. Check result
# -----------------------------
print("\nOriginal vs Imputed counts:\n")
print("Before:")
print(X[X['fuelType'].notna()]['fuelType'].value_counts())

print("\nImputed values added:")
print(pd.Series(fuel_imputed[missing_mask]).map(inv_fuel_mapping).value_counts())

print("\nAfter (fuelType_imputed):")
print(X['fuelType_imputed'].value_counts())

print("\nMissing values in fuelType_imputed:", X['fuelType_imputed'].isna().sum())
```
