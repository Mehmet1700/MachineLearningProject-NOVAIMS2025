## Notebooks
- **(1) 00_data_exploration:** This notebook performs exploratory data analysis on the Cars 4 You dataset to understand its structure, assess data quality, explore feature distributions and relationships with price, and identify insights that will guide preprocessing and modeling decisions.
- **(2) 11_data_mapping_and_normalization:** This notebook performs rule-based data preprocessing, including mapping, normalization, and missing-value handling.
- **(3) ML_Group21_Notebook:** This notebook implements an end-to-end machine learning workflow using Scikit-learn Pipelines, covering all stages from data preprocessing to model evaluation. Multiple regression models introduced during the practical sessions are benchmarked using cross-validation, enabling a consistent and fair comparison of performance and informed model selection. The workflow also incorporates feature selection and hyperparameter tuning to optimize performance, followed by a final evaluation on a holdout set to assess generalization, with the selected model deployed for Kaggle submission. 
- **(4) 122030_model_per_brand:** This notebook explores brand-specific modeling to account for structural price differences between car brands, evaluates multiple tree-based algorithms and a hybrid ensemble, and compares their performance against a single global model to assess the impact of brand-level specialization.
- **(5) 122030_compare_model:** This notebook trains and evaluates the same set of tree-based models and hybrid ensemble as the brand-specific approach, but using a single global model across all brands, serving as a baseline for direct comparison against the brand-level modeling strategy.

## Abstract
The Cars 4 You project addresses the challenge of predicting car prices for an online car resale platform by using machine learning techniques. As the company’s growth has led to increasing delays in manual vehicle inspections, the objective of this work is to develop a reliable predictive model capable of estimating car prices based solely on user-provided information, thereby replacing the mechanic while maintaining pricing accuracy.

Using a subset of Cars 4 You’s historical database from 2020, the project follows a structured end-to-end machine learning workflow. The process begins with exploratory data analysis to understand data quality, feature distributions, and relationships with the target variable, price. This is followed by a rule-based data preprocessing stage, including mapping, normalization, missing-value handling, and encoding preparation. Multiple regression models covered in the course are then benchmarked using consistent evaluation strategies, including cross-validation and holdout validation, to assess their generalization performance.

To improve predictive accuracy, model optimization techniques such as feature selection and hyperparameter tuning are applied. An open-ended analysis further explores whether brand-specific models outperform a single global model, highlighting the role of brand-level heterogeneity in car pricing. Additional large-scale experiments using tree-based models and neural networks are conducted on a SLURM-managed high-performance computing system to extend the analysis beyond standard notebook execution.

The results show that ensemble-based tree models achieve the strongest performance, with brand-specific approaches offering measurable improvements over global models.
